## 1. Tournament Performance Analysis

My agent, Two-Strike Guard, finished first out of 23 agents in the round-robin tournament with 11,937 points. It won 92 matches, had a 69.7% win rate, and earned an average of 179 points per match. Its bracket performance was completely different. Two-Strike Guard tied King kong 30-30 in the first round, but King kong advanced, so my agent was eliminated.

This difference showed me that one result does not always represent the overall quality of an agent. In the bracket, my agent only had one chance to move forward. In the round-robin, every agent played every other agent three times, giving Two-Strike Guard more opportunities to show how consistently it could earn points. Its first-place finish after three complete round-robin tournaments showed that the first-round bracket result did not represent its performance against the full group.

## 2. Strategy Analysis

The top three round-robin agents were Two-Strike Guard, Stable Cooperator, and My Evolved Agent. Two-Strike Guard’s genes were [1.0, 1.0, 0.824421366268778, 0.0, 0.6326476140264558, 0.3448205781863124, 0.0, 0.0]. In plain language, my agent begins by investing and continues cooperating with opponents that cooperate. It also watches recent behavior and protects itself when defection becomes a repeated pattern. This is why I named it Two-Strike Guard: it does not treat one unexpected action as enough reason to abandon cooperation, but it does respond to repeated undercutting.

Stable Cooperator finished second with 11,278 points. Its file describes it as an Always Cooperate strategy that always invests. It only had 15 wins and an 11.4% win rate, but it still earned the second-highest total score. This shows how mutual investment can create a strong overall result even when the agent does not defeat many opponents directly.

My Evolved Agent finished third with 10,997 points and a 79.5% win rate. Its genes show that it also begins cooperatively and responds positively to cooperation, although its memory and response settings differ from mine. The top three agents were not identical, but all three allowed cooperation. Two-Strike Guard performed best because it combined the value of cooperation with a rule for responding to repeated defection.

## 3. Bracket vs. Round-Robin

The bracket champion was Adaptive Survivor, while the round-robin champion was Two-Strike Guard. Adaptive Survivor won 97% of its matches but placed eighth in the round-robin with 9,542 points. Two-Strike Guard had a lower win rate of 69.7% but earned 2,395 more points.

This taught me that the evaluation method affects which agent appears to be the best. Single elimination rewards surviving one matchup at a time, but one tie or difficult opponent can end an agent’s run. The round-robin measures performance across the complete field and reduces the effect of one unusual result. I think the round-robin better reflects overall agent quality because every agent faces the same opponents multiple times. The bracket is useful for measuring head-to-head survival, but the round-robin provides stronger evidence of consistency.

## 4. Superlative Awards and Strategic Trade-offs

Two-Strike Guard received the Most Cooperative award and had the highest average score at 179 points per match. This result fits its decision rules because cooperation is its starting position, while punishment is used as protection rather than as its main behavior.

Each strategy has a trade-off. A highly aggressive agent can win many matchups, but it may cause repeated mutual undercutting, which gives both agents only one point per round. A pure cooperator can earn three points per round through mutual investment but may give an undercutting opponent five points. An agent that punishes too quickly may protect itself, but it can also destroy cooperation after one mistake.

For this tournament, I would rather have a high total score than only a high win rate because the final ranking was based on accumulated points. Adaptive Survivor won more often and became the bracket champion, but Two-Strike Guard earned the highest total score. This showed me that winning individual matchups and producing the most value over time are related, but they are not the same goal.

## 5. Generalization Lesson

Unit 5 began with 10 randomly selected opponents, and I later trained and evaluated the improved version using the broader opponent pool available in that notebook. Most of the Unit 6 student agents were still new competitors with different names, genes, and strategies. Two-Strike Guard could not depend on recognizing a specific opponent, so it had to respond to the actions it observed during each game.

Finishing first suggests that the strategy generalized well to the new environment. Its greatest strength was earning steady points with cooperative agents while keeping some protection against repeated defection. One weakness in this tournament format was that my agent did not earn a clear first-round victory, and its 30–30 tie was still enough to eliminate it. If I could redesign it for this environment, I would add a better early test of the opponent and evaluate future versions using total score, match wins, and bracket survival together. I would not remove its cooperative foundation because that was a major reason it earned the highest score.

## 6. Game Theory and Prescriptive AI Insights

I do not think there is one strategy that will beat every possible opponent. A strategy’s success depends on the strategies around it. Cooperation works when another agent returns it, aggression can exploit an overly trusting agent, and punishment can stop exploitation. However, each advantage can become a weakness in a different matchup.

This same problem appears when businesses choose between investing in a market and starting a price war. Undercutting may create a quick advantage, but repeated undercutting can reduce profits for everyone. Similar choices also appear in supplier relationships and long-term business partnerships, where trust creates value but companies still need protection against unfair behavior.

I now understand this tournament as a prescriptive AI system because the agent does more than describe what its opponent has done. It uses observed behavior, evolved genes, and a defined objective to choose the next action that may improve its future results. Evolutionary optimization did not produce a universally perfect agent. It produced a strategy that performed well under the rewards, competitors, and measurements in this environment. The biggest lesson I learned is that both the training environment and the definition of success shape the strategy that an AI system learns.
